# Adverse Impact Analysis (Lay-friendly)

This notebook demonstrates how to compute **Adverse Impact Ratio (AIR)**.

AIR compares selection rates:
- selection rate among protected candidates
- selection rate among non-protected candidates

AIR = 1.0 means parity (equal selection rates).


```python
import numpy as np
import pandas as pd
import sys
sys.path.append('../02_standards_aligned_pipeline')
import scoring_function as sf

# Toy data example (replace with your real validation data)
rng = np.random.default_rng(42)
n = 2000
df = pd.DataFrame({
    'Protected_Group': rng.integers(0, 2, size=n),
    'High_Performer': rng.binomial(1, 0.30, size=n),
    'Retained': rng.binomial(1, 0.55, size=n),
})
df['score'] = 0.8*df['High_Performer'] + 0.8*df['Retained'] + rng.normal(0, 0.5, size=n)
threshold = np.quantile(df['score'], 0.5)
df['Hire'] = (df['score'] >= threshold).astype(int)

sf.summarize_policy(df['Hire'], df['High_Performer'], df['Retained'], df['Protected_Group'])
```
